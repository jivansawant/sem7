```python
dna = "ATGCGCATGAAATAGCCCTAA"

print(f"len:{len(dna)}")
print(f"dna :{dna.upper()}")

motif="ATG"

for i in range(0,len(dna)-2,1):
    if dna[i:i+3]==motif:
        print(f"motif at {i+1}")

gc_count=dna.count("G")+dna.count("C")
print(f" GC content: {(gc_count/len(dna))*100:.2f}%")

start_codon="ATG"
stop_codons=["TAG","TAA","TGA"]

for i in range(0,len(dna)-1,1):
    if dna[i:i+3]==start_codon:
        for j in range (i+3, len(dna)-2,3):
            codon=dna[j:j+3]
            if codon in stop_codons:
                print(f"coding region :{dna[i:j+3]}")
                break

```

```python
dna = "ATGCGCATGAAATAGCCCTAA"
print(len(dna))

motif="ATG"

for i in range(0,len(dna)-2,1):
    if dna[i:i+3]==motif:
        print(f"motif at {i+1}")

gc=dna.count('G')+dna.count('C')
print(f"Gc percent :{round((gc/len(dna)*100),2)}%")

start_codons=['ATG']
stop_codons=["TAG","TGA"]

for i in range(0,len(dna)-1,1):
    if dna[i:i+3] in start_codons:
        for j in range(i+3,len(dna)-2,3):
            c=dna[j:j+3]
            if c in stop_codons:
                print(f"coding region {dna[i:j+3]}")
                break

```
