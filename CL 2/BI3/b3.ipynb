```python

# Practical: Protein Structure Prediction

import warnings
warnings.filterwarnings("ignore")

# Step 1: Validate protein sequence
def validate_sequence(sequence):

    valid_amino_acids = "ACDEFGHIKLMNPQRSTVWY"

    for amino_acid in sequence:
        if amino_acid not in valid_amino_acids:
            return False

    return True


# Step 2: Calculate sequence identity
def sequence_identity(target, template):

    length = min(len(target), len(template))

    if length == 0:
        return 0

    matches = 0

    for i in range(length):
        if target[i] == template[i]:
            matches += 1

    identity = (matches / length) * 100

    return identity


# Step 3: Predict simple secondary structure
def predict_secondary_structure(sequence):

    helix = "ALMEKQRH"
    sheet = "VIFYWT"

    structure = ""

    for amino_acid in sequence:

        if amino_acid in helix:
            structure += "H"

        elif amino_acid in sheet:
            structure += "E"

        else:
            structure += "C"

    return structure


# Step 4: Define protein sequence
sequence = "MKTLLVAGAAAGTASAAPVLAADKPEMVRRLAEAGVEVVVEPGDNVVAMRDPQT"


# Step 5: Validate sequence
if not validate_sequence(sequence):

    print("Invalid protein sequence.")

else:

    print("\nProtein sequence is valid.")
    print("Protein sequence:", sequence)
    print("Protein length:", len(sequence), "amino acids")


    # Step 6: Define template sequence
    template = "MKTIIALSYIFCLVFADYKDDDDKGQRLIFAGKQLEDGRTLSDYVAVIDALNDSD"

    print("\nTemplate sequence:")
    print(template)


    # Step 7: Calculate sequence identity
    identity = sequence_identity(sequence, template)

    print("\nSequence Identity:",
          round(identity, 2), "%")


    # Step 8: Predict secondary structure
    structure = predict_secondary_structure(sequence)

    print("\nSecondary Structure:")
    print(structure)


    # Step 9: Estimate confidence
    if identity >= 50:
        confidence = "High"

    elif identity >= 30:
        confidence = "Moderate"

    else:
        confidence = "Low"

    print("\nModeling Confidence:", confidence)


    # Step 10: Display structural representation
    print("\nPosition  Amino Acid  Structure")

    for i in range(len(sequence)):

        if structure[i] == "H":
            name = "Alpha Helix"

        elif structure[i] == "E":
            name = "Beta Sheet"

        else:
            name = "Coil"

        print(i + 1, "       ",
              sequence[i], "          ", name)


    # Step 11: Final result
    print("\nProtein Structure Analysis Completed.")
```
