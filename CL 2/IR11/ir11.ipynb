```python
# PageRank Algorithm

# Step 1: Define the web pages and their outgoing links
graph = {
    "A": ["B", "C"],
    "B": ["C"],
    "C": ["A"],
    "D": ["C"]
}

# Step 2: Calculate the total number of pages
pages = list(graph.keys())
N = len(pages)

# Step 3: Set the damping factor
damping = 0.85

# Step 4: Give each page an initial PageRank score
ranks = {}

for page in pages:
    ranks[page] = 1 / N

# Step 5: Repeat the calculation 100 times
for iteration in range(100):

    new_ranks = {}

    # Initialize the new scores
    for page in pages:
        new_ranks[page] = (1 - damping) / N

    # Distribute scores through outgoing links
    for source in pages:
        for target in graph[source]:
            new_ranks[target] = (
                new_ranks[target]
                + damping * ranks[source] / len(graph[source])
            )

    # Update the PageRank scores
    ranks = new_ranks

# Step 6: Display the final PageRank scores
print("PageRank Scores:")

for page in pages:
    print(page, ":", round(ranks[page], 4))

# Step 7: Display pages in descending order of importance
print("\nPages ranked by importance:")

sorted_pages = sorted(
    ranks,
    key=ranks.get,
    reverse=True
)

for page in sorted_pages:
    print(page, ":", round(ranks[page], 4))
```
