```python

import pandas as pd
from scipy.stats import ttest_ind

# 1. Create RNA-Seq dataset
df = pd.DataFrame({
    "Gene": ["Gene_A", "Gene_B", "Gene_C", "Gene_D", "Gene_E"],
    "Control1": [100, 200, 150, 80, 300],
    "Control2": [110, 190, 160, 85, 290],
    "Control3": [105, 210, 155, 75, 310],
    "Treatment1": [400, 180, 145, 300, 305],
    "Treatment2": [420, 195, 150, 310, 295],
    "Treatment3": [390, 205, 158, 290, 315]
})

print("RNA-Seq Dataset:")
print(df)

# 2. Compare gene expression
results = []

for i in range(len(df)):

    control = df.iloc[i, 1:4].astype(float)
    treatment = df.iloc[i, 4:7].astype(float)

    p_value = ttest_ind(treatment, control).pvalue

    fold_change = treatment.mean() / control.mean()

    if p_value < 0.05 and fold_change >= 2:
        status = "Upregulated"
    elif p_value < 0.05 and fold_change <= 0.5:
        status = "Downregulated"
    else:
        status = "Not Significant"

    results.append([df["Gene"][i], fold_change, p_value, status])

# 3. Display results
result_df = pd.DataFrame(
    results,
    columns=["Gene", "Fold Change", "P Value", "Status"]
)

print("\nDifferential Gene Expression Results:")
print(result_df.round(4).to_string(index=False))
```
