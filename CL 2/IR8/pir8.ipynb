```python
documents={
    1:"apple banana orange",
    2:"banana orange",
    3:"apple banana mango"

}
print("documents:", documents)

inverted_index={}

for doc_id , docs in documents.items():
    words=docs.split()

    for word in words:
        if word not in inverted_index:
            inverted_index[word]=[]
        if doc_id not in inverted_index[word]:
            inverted_index[word].append(doc_id)

print(f"invetred_index :{inverted_index}")

query="mango"

if query in inverted_index:
    print(f"dpucments of {query} are : {inverted_index[query]}")
else:
    print(f"{query} results not found")

```
